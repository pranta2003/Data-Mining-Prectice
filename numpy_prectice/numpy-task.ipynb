```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Student': ['Akash', 'Bithi', 'Choyon', 'Disha'],
    'Attendance': ['R','NR','R','R'],
    'Gender': ['M','F','M','F'],
    'Math': [80, 82, 40, 20],
    'CSE': [70, 72, 50, 25],
    'English': [90, 88, 45, 15],
    'Grade': ['A','A','B', 'C']
})  

print(df)
```

```python
# Find the similarity matrix of numeric attributes
# Find the dissimilarity matrix of Nominal attributes
# Find the dissimilarity matrix of binary attributes
# Find the dissimilarity matrix of ordinal attributes
```

```python
#1
f = df[['Math' , 'CSE' , 'English']].values

ds = np.zeros((len(f), len(f)))

for i in range (len(f)):
 for j in range (len(f)):
     ds[i][j] = np.sum(np.abs(f[i]-f[j]))


 print(ds)


```

```python
#2


c = ['Attendance', 'Gender', ]
f = df[c].values


p = len(c)
n = len(f)


nd = np.zeros((n, n))

for i in range(n):
    for j in range(n):
 
        m = np.sum(f[i] == f[j])
        nd[i, j] = (p - m) / p

print("Nominal Dissimilarity Matrix:\n", nd )

```

```python
#3



g = {'C': 3, 'B': 2, 'A': 1}
ranks = df['Grade'].str.strip().map(g).values

small = min(g.values())
M = max(g.values())
z = (ranks - small) / (M - small)

n = len(z)
ordinal_dissimilarity = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        ordinal_dissimilarity[i, j] = np.abs(z[i] - z[j])

print(ordinal_dissimilarity)
```

```python
import pandas as pd
import numpy as np

# Select binary attributes
bi = df[['Attendance', 'Gender']].copy()

# Convert binary values to 0 and 1
bi['Attendance'] = bi['Attendance'].map({
    'R': 1,
    'NR': 0
})

bi['Gender'] = bi['Gender'].map({
    'M': 1,
    'F': 0
})

# Create an empty dissimilarity matrix
n = len(bi)
ds = np.zeros((n, n))

# Calculate dissimilarity
for i in range(n):
    for j in range(n):
        differences = np.sum(bi.iloc[i] != bi.iloc[j])
        ds[i][j] = differences / len(bi.columns)

# Display the result
print("Binary Dissimilarity Matrix:")
print(ds)
```
