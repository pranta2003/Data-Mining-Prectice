```python


m = np.array([55 , 60 , 72 , 88 , 85 , 90 , 95])

Q1 = np.percentile(m, 25)
Q3 = np.percentile(m, 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = m[(m < lower) | (m > upper)]

print(outliers)
```
