```python


m = np.array([55 , 60 , 72 , 88 , 85 , 90 , 95])

Q1 = np.percentile(m, 25)
Q3 = np.percentile(m, 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = m[(m < lower) | (m > upper)]

print(outliers)
```

```python
#Manhattan

import numpy as np


dissimilarityM = np.zeros((len(f), len(f)))

for i in range(len(f)):
  for j in range(len(f)):
    dissimilarityM[i][j] = np.sum(np.abs(f[i]-f[j]))
      
    
print(dissimilarityM)

```

```python
#Eucleadian


dissimilarityEucleadian = np.zeros((len(f), len(f)))

for i in range(len(f)):
  for j in range(len(f)):
    dissimilarityEucleadian[i][j] = np.sqrt(np.sum(f[i]-f[j]) ** 2)
      
    
print(dissimilarityEucleadian)
```

```python

ia = pd.DataFrame({
    'Student' : [ 'A' , 'B ', 'C' , 'D'],
    'Math'    : [ 62 , 52 , 87 , 92 ],
    'CSE'     : [ 45 , 85 , 98 , 48 ] ,
    'English' : [ 75 , 85 , 78 , 58 ] ,
    'Grade'   : [ 'A' ,'A' , 'B ' , 'C' ] 
}) 


c = ['Math', 'CSE', 'English']
f = ia[c].values


p = len(c)
n = len(f)


nominal_dissimilarity = np.zeros((n, n))

for i in range(n):
    for j in range(n):
 
        m = np.sum(f[i] == f[j])
        nominal_dissimilarity[i, j] = (p - m) / p

print("Nominal Dissimilarity Matrix:\n", nominal_dissimilarity )
print()
```

```python


ia = pd.DataFrame({
    'Student' : [ 'A' , 'B ', 'C' , 'D'],
    'Math'    : [ 62 , 52 , 87 , 92 ],
    'CSE'     : [ 45 , 85 , 98 , 48 ] ,
    'English' : [ 75 , 85 , 78 , 58 ] ,
    'Grade'   : [ 'A' ,'A' , 'B ' , 'C' ] 
}) 



g = {'C': 1, 'B': 2, 'A': 3}
ranks = ia['Grade'].str.strip().map(g).values

small = min(g.values())
M = max(g.values())
z = (ranks - small) / (M - small)

n = len(z)
ordinal_dissimilarity = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        ordinal_dissimilarity[i, j] = np.abs(z[i] - z[j])

print(ordinal_dissimilarity)
```

```python
print("Nominal similarity Matrix:\n", 1-nominal_dissimilarity )
```
